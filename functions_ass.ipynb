```python
# Creating daily sales for 7 days
sales = []
for day in range (1,8):
    amount = float(input(f'Enter sales for the day {day}:'))
    sales.append(amount)
```

```python
# Calculating total sales

total_sales = sum(sales)
print(total_sales)
```

```python
# Calculating total average sales
average_sales = total_sales/7
print(average_sales)
```

```python
# Rating performance based on average
def evaluate_sales(sales):
    total_sales = sum(sales)
    average_sales = total_sales/len(sales)
    if average_sales >= 80000:
        performance = "Excellent"
    elif average_sales >= 50000:
        performance = "Good"
    else: 
        performance = "Needs improvement"
   
    return total_sales,average_sales,performance
```

```python
evaluate_sales(sales)
```

```python
type(evaluate_sales(sales))
```

```python
total,average,performance = evaluate_sales(sales)
print(f"Total sales: NGN {total:,}")
print(f"Average sales: NGN {average:,.2f}")
print(f"Performance Rating: {performance}")
```
