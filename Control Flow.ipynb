```python
score = 45
if score >= 80:
    print(f" Your score {score} is A")
elif score >= 70:
    print(f" Your score {score} is B")
elif score >= 60:
    print(f"Your score {score} is C")
elif score >= 50:
    print(f"Your score {score} is D")
else:
    print(f"Your score {score} is F")
```
